```python
# Q35
import copy

a = [[1, 2], [3, 4]]
b = copy.copy(a)
b[0][1] = 99
print(a)
# [[1, 99], [3, 4]]
# b is a shallow copy of a so changing elements in b also changes elements in a as well.
```

```python
# Q36
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
#  [[1, 2], [3, 4]]
# Deep copy just copies the elements so changes are not carried forward
```
