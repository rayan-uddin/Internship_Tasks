```python
# Q1
def count_of_words(self):
    words = self.split()

    words_dict = dict()
    for i in words:
        count = 0
        for j in words:
            if i == j:
                count += 1
                words_dict[i] = count

    return words_dict
```

```python
# Q3

def appears_once_word(self):
    for i in self:
        count = 0

        for j in self:
            if i.lower() == j.lower():
                count += 1

        if count == 1:
            return i

    return None
```

```python
# Q7 

def second_largest_distinct_number(l1):
    i = 0
    unique_number = []

    while i < len(l1):
        if l1[i] not in unique_number:
            unique_number.append(l1[i])
        else:
            i += 1

    if len(unique_number) == 1:
        return None

    unique_number.sort()

    second_largest_number = unique_number[1]
    
    return second_largest_number
```

```python

# Q13

def move_by_k(list, k):
    n = len(list)
    k = k % n
    result = []
    
    for i in range(n):
        result.append(list[(i - k) % n])
    
    return result
```

```python
# Q14

def highest_sum_cons_num(list):
    highest_sum = 0
    for i in range(len(list) - 1):
        if int(list[i]) + int(list[i+1]) > highest_sum:
            highest_sum = int(list[i]) + int(list[i + 1])
    
    return highest_sum
```

2. creating a new list
3. for looping in range (len(list)) and again looping in tange(i + 1, len(list)) -> starting from i + 1 to avoid duplicates
4. if the sum of both list[i] and list[j] equals to target
5. then append in the new list but keep the sorted
6. now sort the ewn list but using the condition the only the first value is taken into consideration

```python
# Q15

def unique_pair_sum(nums, target):
    sequences = []

    for i in range(len(nums)):
        for j in range(i+1, len(nums)):
            if nums[i] + nums[j] == target:
                pair = sorted([nums[i], nums[j]])
                sequences.append(pair)

    sequences.sort(key = lambda x: x[0])
    return sequences
```
