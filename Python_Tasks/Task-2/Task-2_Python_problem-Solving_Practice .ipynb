Q1. Find the Shortest Word in a Sentence

Write a Python program to find the shortest word in a given sentence.

Example input:
Python makes coding easy
Expected output:
easy

## Approach:
```
Taking input
splitting the input to create a list
taking length of first word to be the smallest (Hypothetical)
checking throught the list for a word having length smaller than the hypothetical if present change them
fianlly print the final output
```

````python
sentence = input("Enter Input Sentence: ")
# sentence = "Python makes coding easy"
words = sentence.split()
shortest_word = words[0]

for word in words:
    if len(word) < len(shortest_word):
        shortest_word = word

print("Given sentence: ", sentence)
print(shortest_word)
````

Q6. Count Vowels in a String

Write a Python program to count the total number of vowels in a string. Consider both uppercase and lowercase vowels.

Q8. Find the Most Frequent Character

Write a Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, return the first one encountered. Ignore spaces.

Example input:
banana
Expected output:
a

```
assinging max frq as 0 and max count as 0
looping through the word for eacxh letter
then again looping inside the loop for going through every char in the word
counting each word which repeats and increasing the counter
finally retunring the word with the highest count
```

````python
word = "banana"

max_count = 0
max_freq = ""
for i in word:
    count = 0

    for j in word:
        if j == i:
            count += 1

    if count > max_count:
        max_freq = j
        max_count = count

print(max_freq, "Number of Count: ", max_count)
````

Q16. Find the Most Frequent Word in a Sentence


Write a Python program to find the word that appears most frequently in a sentence. Treat uppercase and lowercase versions of a word as the same word. If multiple words share the highest frequency, return the one that appears first.

Example input:
Python is easy and Python is powerful
Expected output:
python

```
SAME AS PREVIOUS QUESTION
```

````python
sentence = "Python is easy and Python is powerful"
words = sentence.split()

max_count = 0
max_freq = 0

for i in words:
    count = 0

    for j in words:
        if j.lower() == i.lower():
            count += 1

    if count > max_count:
        max_freq = i.lower()
        max_count = count

print(max_freq, max_count)
````

Q17. Find All Words Appearing More Than Once


Write a Python program to find all words that appear more than once in a sentence. Display each repeated word only once, in the order of its first appearance. Treat uppercase and lowercase versions as the same word.


Example input:
data science uses data and science tools
Expected output:
['data', 'science']

```
taking input and creating a list using .split()
taking max_count as 1 and an empty list for repeated_words
looping through the words
then looping again
and incrementing count by 1 if words.lower() are equals
appending into lsit and giving output
```

````python
sentence = "data science uses data and science tools"
words = sentence.split()

max_count = 1
repeated_words = []

for i in words:
    count = 0

    for j in words:
        if i == j:
            count += 1

    if count > max_count and i not in repeated_words:
        repeated_words.append(i)

repeated_words
````

Q18. Find the First Word That Appears Only Once


Write a Python program to find the first word that occurs exactly once in a sentence. Treat uppercase and lowercase versions as the same word.


Example input:
apple banana apple mango banana orange
Expected output:
mango

```
taking list as input
taking max_count as 2 bcuz need word not repeating 
looping twice and incrementing the count for equal values
returning non repeated words list
```

````python
sentence = "apple banana apple mango banana orange"
words = sentence.split()

max_count = 2
non_repeated_words = []


for i in words:
    count = 0

    for j in words:
        if i.lower() == j.lower():
            count += 1

    if count < max_count and i.lower() not in non_repeated_words:
        non_repeated_words.append(i.lower())

print(non_repeated_words)
````

Q20. Find the Longest Word That Appears Only Once

Write a Python program to find the longest word that appears exactly once in a sentence. If multiple qualifying words have the same maximum length, return the first one encountered. Treat uppercase and lowercase versions as the same word.

Example input:
Python makes coding useful Python
Expected output:
coding

```
Same as Q18 but also taking len() into consideration
```

````python
sentence = "Python makes coding useful Python"
words = sentence.split()
longest_word = ""

for i in words:
    count = 0

    for j in words:
        if i.lower() == j.lower():
            count += 1

    if count == 1 and len(i) > len(longest_word):
        longest_word = i

print(longest_word)
````
